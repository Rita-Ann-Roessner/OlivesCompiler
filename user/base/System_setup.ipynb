```python
# apo system
u1 = mda.Universe('receptor_cg.pdb')

# all atom receptor + ghrelin peptide
u2 = mda.Universe('/mnt/DATA20To/rita/240614_msm_3.0/7f9y/protein.gro')
# ghrelin peptie as new reference
ref = u2.select_atoms('name CA and resid 1 to 15')

u3 = mda.Universe('ghrelin_cg.pdb') # alpha-model of octanyolated ghrelin peptide
mobile = u3.select_atoms('name CA and resid 1 to 15')

# centers from the fitting selections
mob_com = mobile.center_of_mass()
ref_com = ref.center_of_mass()

# rotation from centered coordinates of the fitting atoms
R, rmsd = align.rotation_matrix(mobile.positions - mob_com,
                                ref.positions    - ref_com)

u3.atoms.translate(-mob_com)
u3.atoms.rotate(R)
u3.atoms.translate(ref_com)
sel = u3.select_atoms('all')
sel.write('ghrelin-aligned.pdb')

# add aligned ghrelin peptide to apo system
u3.residues.resids += u1.residues.resids.max() + 1
merged = mda.Merge(u1.atoms, u3.atoms)

merged.atoms.write('init_cg.pdb')
```

```python
# adjust topology
top_path = Path("system.top")
text = top_path.read_text()

# ---- 1) include + molecules inserts (your existing logic)
has_itp = re.search(r'^\s*#include\s+"molecule_1\.itp"\s*$', text, flags=re.M)
has_mol = re.search(r'^\s*molecule_1\s+1\s*$', text, flags=re.M)

new_text = text

if not has_itp:
    new_text = re.sub(
        r'(^\s*#include\s+"molecule_0\.itp"\s*$)',
        r'\1\n#include "molecule_1.itp"',
        new_text,
        count=1,
        flags=re.M
    )

if not has_mol:
    new_text = re.sub(
        r'(^\s*molecule_0\s+1\s*$)',
        r'\1\nmolecule_1          1',
        new_text,
        count=1,
        flags=re.M
    )

# ---- 2) REPLACE martini include line
# Only replace if not already pointing to the _Up_031c file
has_new_martini = re.search(
    r'^\s*#include\s+"martini_v3\.0\.0_Up_031c\.itp"\s*$',
    new_text, flags=re.M
)

if not has_new_martini:
    # Replace the exact martini_v3.0.0.itp include, preserving leading spaces
    new_text, n_repl = re.subn(
        r'^(\s*)#include\s+"martini_v3\.0\.0\.itp"\s*$',
        r'\1#include "martini_v3.0.0_Up_031c.itp"',
        new_text,
        count=1,
        flags=re.M
    )
    # (optional) sanity check if nothing was replaced:
    if n_repl == 0:
        print("Warning: did not find '#include \"martini_v3.0.0.itp\"' to replace.")

# ---- 3) write back if changed (with backup)
if new_text != text:
    top_path.with_suffix(".top.bak").write_text(text)
    top_path.write_text(new_text)
    print("system.top updated (backup: system.top.bak)")
else:
    print("No changes needed.")
```
